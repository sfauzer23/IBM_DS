### Pre-processing

```python
# Import necessary libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
```

```python
# Import data

ilepath = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/medical_insurance_dataset.csv'
df = pd.read_csv(filepath, header = None)
```

```python
# Set dataframe data
headers = [
    "age",
    "gender",
    "bmi",
    "no_of_children",
    "smoker",
    "region",
    "charges"
]

df.columns = headers
```

```python
# replace ? with nan

df.replace("?", np.nan, inplace = True)
```

### Data Wrangling

```python
# Here we can see that column smoker and age has Nulls

df.info()
```

```python
# To deal with missing values:
# For continuous attributes (e.g., age), replace missing values with the mean.
# For categorical attributes (e.g., smoker), replace missing values with the most frequent value.
```

```python
# Dealing with age Nulls

avg_age = df["age"].astype("float").mean(axis=0)
df["age"].replace(np.nan, avg_age, inplace=True)
```

```python
# Dealing with Smoker

freq = df["smoker"].mode().iloc[0]
df["smoker"].replace(np.nan, freq, inplace = True)
```

```python
# Update the datatypes

df[["age", "smoker"]] = df[["age", "smoker"]].astype("int")
```

```python
# Change row charges to round to two decimal places

df[["charges"]] = np.round(df[["charges"]], 2)
```

### Exploratory Data Analysis

```python
# Regression Plot


width = 12
height = 10
plt.figure(figsize=(width, height))
sns.regplot(x="bmi", y="charges", data=df)
plt.ylim(0,)
```

```python
# Box Plot

width = 12
height = 10
plt.figure(figsize=(width, height))
sns.boxplot(x="smoker", y="charges", data=df)
plt.ylim(0,)
```

### Model Development

```python
# Setup the linear regression object

lm = LinearRegression()
lm
```

```python
# Define Variables

X = df[['smoker']]
Y = df['charges']
```

```python
# Fit data

lm.fit(X,Y)
```

```python
# R2

lm.score(X, Y)
```

```python
lm2 = LinearRegression()
lm2
```

```python
# Define Variables

X2 = df[['smoker', "age", "bmi", "no_of_children", "region"]]
Y2 = df['charges']
```

```python
# Fit data

lm.fit(X2,Y2)
```

```python
# R2

lm.score(X2, Y2)
```

```python
# Pipeline

Input=[('scale',StandardScaler()), ('polynomial', PolynomialFeatures(include_bias=False)), ('model',LinearRegression())]

y = df['charges']
```

```python
pipe=Pipeline(Input)
pipe
```

```python
X2 = X2.astype(float)
pipe.fit(X2,y)
```

```python
# R2

y_pipe = pipe.predict(X2)
r2_score(y, y_pipe)
```

### Model Refinement

```python
# Split up the data
x_train, x_test, y_train, y_test = train_test_split(X2, Y2, test_size=0.20, random_state=1)
```

```python
# Ridge regression

RigeModel=Ridge(alpha = 0.1)
RigeModel.fit(x_train, y_train)

yhat = RigeModel.predict(x_test)
r2_score(y_test, yhat)
```

```python
# Doing a ridge with a poylnomial transformation

pr = PolynomialFeatures(degree=2)
x_train_pr = pr.fit_transform(x_train)
x_test_pr = pr.transform(x_test)
RigeModel.fit(x_train_pr, y_train)
y_hat = RigeModel.predict(x_test_pr)
print(r2_score(y_test,y_hat))
```
